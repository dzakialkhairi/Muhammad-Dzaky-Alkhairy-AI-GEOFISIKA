## **HASIL**

```python
import numpy as np
import matplotlib.pyplot as plt

kedalaman = [0, 100, 400, 600]
resistivitas = [12.5, 65, 220]

fig, ax = plt.subplots(figsize=(9, 6))

warna = ["#2b83ba", "#abdda4", "#d7191c"]

for i in range(3):
    ax.fill_between(
        [0, 1000],
        kedalaman[i],
        kedalaman[i + 1],
        color=warna[i]
    )

ax.invert_yaxis()
ax.set_xlabel("Jarak horizontal (m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Model Konseptual Sistem Panas Bumi")

ax.text(500, 50, "Cap rock (dugaan)\n5–20 Ωm",
        ha="center", va="center")
ax.text(500, 250, "Reservoir (dugaan)\n30–100 Ωm",
        ha="center", va="center")
ax.text(500, 500, "Batuan resistif\n>200 Ωm",
        ha="center", va="center")

ax.set_xlim(0, 1000)
ax.set_ylim(600, 0)
ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

Berdasarkan data resistivitas, sistem panas bumi pada kawasan vulkanik aktif dapat diinterpretasikan menjadi tiga zona utama, yaitu zona alterasi, cap rock, dan reservoir. Pada kedalaman 0–100 m, nilai resistivitas sebesar 5–20 Ωm mengindikasikan material konduktif yang dapat berkaitan dengan batuan teralterasi hidrotermal, lempung, atau sedimen jenuh air. Zona ini berpotensi merepresentasikan bagian dari cap rock, terutama apabila mineral lempung hasil alterasi hidrotermal mendominasi.

Pada kedalaman 100–400 m, resistivitas meningkat menjadi 30–100 Ωm. Nilai tersebut dapat mengindikasikan batuan dengan tingkat alterasi lebih rendah, rekahan yang terisi fluida, atau batuan reservoir yang memiliki permeabilitas memadai. Zona ini berpotensi menjadi reservoir panas bumi apabila didukung oleh keberadaan fluida panas, porositas sekunder, dan struktur geologi permeabel.

Sementara itu, pada kedalaman lebih dari 400 m, resistivitas di atas 200 Ωm dapat menunjukkan batuan dasar yang relatif kompak dan memiliki porositas rendah. Namun, resistivitas tinggi juga dapat berkaitan dengan batuan vulkanik segar atau intrusi beku, sehingga tidak secara langsung membuktikan keberadaan sumber panas.

Secara konseptual, model tersebut sesuai dengan sistem panas bumi vulkanik yang memiliki lapisan penudung konduktif dan zona reservoir di bawahnya. Meskipun demikian, penentuan zona secara definitif memerlukan korelasi dengan data geologi, manifestasi permukaan, dan temperatur bawah permukaan.

## **HASIL**

```python

import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# Data Vs berdasarkan input
kedalaman = [0, 5, 15, 30]
vs = [150, 250, 500]

warna = ["#d73027", "#fdae61", "#1a9850"]
label = [
    "Vs = 150 m/s\nRelatif lunak\n0–5 m",
    "Vs = 250 m/s\nKekakuan menengah\n5–15 m",
    "Vs = 500 m/s\nRelatif kaku\n>15 m"
]

fig, ax = plt.subplots(figsize=(9, 6))

for i in range(3):
    ax.fill_betweenx(
        [kedalaman[i], kedalaman[i + 1]],
        0, 1,
        color=warna[i],
        edgecolor="black"
    )

    tengah = (kedalaman[i] + kedalaman[i + 1]) / 2

    ax.text(
        0.5, tengah, label[i],
        ha="center", va="center",
        fontsize=10, fontweight="bold"
    )

ax.invert_yaxis()
ax.set_xlim(0, 1)
ax.set_ylim(30, 0)
ax.set_xticks([])
ax.set_ylabel("Kedalaman (m)")
ax.set_yticks([0, 5, 10, 15, 20, 25, 30])

ax.set_title(
    "Model Konseptual Profil Vs MASW\n"
    "Evaluasi Awal Potensi Ketidakstabilan Lereng"
)

legenda = [
    Patch(facecolor=warna[0], label="Vs rendah: 150 m/s"),
    Patch(facecolor=warna[1], label="Vs menengah: 250 m/s"),
    Patch(facecolor=warna[2], label="Vs tinggi: 500 m/s")
]

ax.legend(handles=legenda, loc="lower right")
ax.grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.savefig("profil_vs_masw.png", dpi=200)
plt.show()
```

Profil Vs menunjukkan variasi kekakuan material terhadap kedalaman. Lapisan permukaan pada kedalaman 0–5 m memiliki Vs sebesar 150 m/s, yang mengindikasikan material relatif lunak dan berpotensi mengalami deformasi lebih besar. Pada lereng curam dengan riwayat longsor, lapisan ini perlu mendapat perhatian karena dapat mengalami penurunan kekuatan ketika jenuh air.

Pada kedalaman 5–15 m, Vs meningkat menjadi 250 m/s, menunjukkan material yang relatif lebih kaku dibandingkan lapisan permukaan. Batas antara kedua lapisan tersebut merupakan perubahan sifat mekanik yang perlu diperiksa lebih lanjut, tetapi belum dapat dinyatakan sebagai bidang gelincir.

Di bawah kedalaman 15 m, Vs sebesar 500 m/s menunjukkan material yang relatif lebih kaku, yang dapat berkaitan dengan tanah sangat padat atau batuan lapuk hingga batuan lebih kompak, bergantung pada kondisi geologi setempat.

Secara keseluruhan, lapisan 0–5 m merupakan zona prioritas untuk pemeriksaan geoteknik, bukan bukti pasti zona longsor. Penentuan bidang gelincir dan kestabilan lereng memerlukan data tambahan seperti geometri lereng, kondisi air tanah, kekuatan geser material, dan diskontinuitas batuan

## **HASIL**

```python

import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# Batas kedalaman ilustratif (m)
depth = [0, 20, 60, 100]

# Rentang resistivitas dari soal
labels = [
    "Lapisan penutup\nResistivitas belum diketahui",
    "Kandidat akuifer\n10–30 Ωm",
    "Lapisan resistif\nSekitar 200 Ωm"
]

colors = ["#bdbdbd", "#2b83ba", "#fdae61"]

fig, ax = plt.subplots(figsize=(9, 6))

for i in range(3):
    ax.fill_betweenx(
        [depth[i], depth[i + 1]],
        0, 1,
        color=colors[i],
        edgecolor="black"
    )

    mid = (depth[i] + depth[i + 1]) / 2
    ax.text(
        0.5, mid, labels[i],
        ha="center", va="center",
        fontsize=10
    )

ax.invert_yaxis()
ax.set_xlim(0, 1)
ax.set_ylim(100, 0)
ax.set_xticks([])
ax.set_ylabel("Kedalaman (m)")
ax.set_yticks([0, 20, 40, 60, 80, 100])

ax.set_title(
    "Model Konseptual Resistivitas untuk Identifikasi Akuifer"
)

legend = [
    Patch(facecolor=colors[0], label="Lapisan penutup: data tidak tersedia"),
    Patch(facecolor=colors[1], label="Kandidat akuifer: 10–30 Ωm"),
    Patch(facecolor=colors[2], label="Lapisan resistif: ~200 Ωm")
]

ax.legend(handles=legend, loc="lower right")
ax.grid(axis="y", alpha=0.2)

plt.tight_layout()
plt.savefig("model_konseptual_akuifer.png", dpi=200)
plt.show()
```

Zona pada kedalaman 20–60 m dengan resistivitas 10–30 Ωm merupakan kandidat zona jenuh air yang perlu diselidiki lebih lanjut. Nilai tersebut juga dapat menunjukkan material lempungan yang memiliki permeabilitas rendah. Lapisan di bawahnya dengan resistivitas sekitar 200 Ωm berpotensi berupa batuan kompak atau material yang relatif kering. Karena itu, keberadaan akuifer belum dapat dipastikan hanya berdasarkan resistivitas.

## **HASIL**

```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm

# Nilai anomali dari soal (nT)
anomali = np.array([
    [-100, -100, 300, 300, 800],
    [-100,  300, 300, 800, 800],
    [ 300,  300, 800, 800, 300],
    [-100,  300, 300, 300, -100],
    [-100, -100, 300, 800, 300]
])

# Kategori warna untuk nilai anomali
cmap = ListedColormap(["#2c7bb6", "#ffffbf", "#d7191c"])
norm = BoundaryNorm([-200, 0, 500, 1000], cmap.N)

fig, ax = plt.subplots(figsize=(8, 6))

im = ax.imshow(
    anomali,
    cmap=cmap,
    norm=norm,
    origin="lower",
    interpolation="nearest",
    aspect="auto"
)

# Anotasi nilai pada setiap sel
for i in range(anomali.shape[0]):
    for j in range(anomali.shape[1]):
        ax.text(
            j, i, f"{anomali[i, j]}",
            ha="center", va="center",
            fontsize=9
        )

cbar = plt.colorbar(im, ax=ax, ticks=[-100, 300, 800])
cbar.set_label("Anomali magnetik (nT)")
cbar.ax.set_yticklabels([
    "Rendah: -100 nT",
    "Sedang: +300 nT",
    "Tinggi: +800 nT"
])

ax.set_title("Model Konseptual Anomali Magnetik")
ax.set_xlabel("Indeks posisi horizontal (ilustratif)")
ax.set_ylabel("Indeks posisi horizontal (ilustratif)")

plt.tight_layout()
plt.savefig("peta_konseptual_anomali_magnetik.png", dpi=200)
plt.show()
```

Anomali magnetik tinggi sebesar +800 nT menunjukkan adanya sumber yang menghasilkan respons magnetik relatif kuat. Kondisi ini dapat berkaitan dengan batuan beku mafik, tubuh intrusi, atau batuan yang mengandung magnetit. Oleh karena itu, zona tersebut dapat diprioritaskan sebagai target awal eksplorasi mineral besi apabila sesuai dengan kondisi geologi regional.

Anomali sedang sebesar +300 nT dapat merepresentasikan batuan dengan sifat kemagnetan menengah atau pengaruh beberapa sumber magnetik. Sementara itu, anomali rendah sebesar −100 nT dapat berkaitan dengan batuan bersuseptibilitas rendah, zona alterasi yang menghancurkan mineral magnetik, atau pengaruh geometri dan arah magnetisasi sumber.

Untuk mineralisasi sulfida, anomali rendah maupun tinggi perlu dievaluasi bersama informasi geologi karena sulfida tidak selalu bersifat magnetik. Dengan demikian, anomali +800 nT merupakan target awal yang menarik, tetapi belum cukup untuk memastikan keberadaan endapan besi maupun sulfida.

```python

import matplotlib.pyplot as plt

# Hasil korelasi dari analisis
faktor = ["Monsun (lag-1)", "ENSO (lag-1)"]
korelasi = [0.577, -0.134]

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.barh(faktor, korelasi)

ax.axvline(0, color="black", linewidth=1)
ax.set_xlim(-0.25, 0.70)
ax.set_xlabel("Koefisien korelasi (r)")
ax.set_title(
    "Korelasi Monsun dan ENSO dengan MHW\n"
    "Laut Banda, 2005–2024"
)
ax.grid(axis="x", alpha=0.25)

for bar, nilai in zip(bars, korelasi):
    posisi = nilai + 0.015 if nilai >= 0 else nilai - 0.015
    perataan = "left" if nilai >= 0 else "right"

    ax.text(
        posisi,
        bar.get_y() + bar.get_height() / 2,
        f"r = {nilai:.3f}",
        va="center",
        ha=perataan
    )

plt.tight_layout()
plt.savefig("korelasi_monsun_enso_mhw.png", dpi=200)
plt.show()
```

Hasil analisis menunjukkan bahwa monsun memiliki korelasi positif yang lebih kuat dengan kejadian Marine Heatwave (MHW) di Laut Banda, dengan koefisien korelasi lag-1 sebesar 0,577. Temuan ini mengindikasikan bahwa variasi monsun memiliki hubungan temporal yang relatif kuat dengan kejadian MHW pada jeda waktu yang dianalisis. Secara fisik, hubungan tersebut dapat berkaitan dengan perubahan kecepatan angin, pencampuran vertikal, dan distribusi panas di lapisan permukaan laut. Namun, mekanisme dominannya perlu diverifikasi menggunakan data angin dan fluks panas.

Sebaliknya, ENSO menunjukkan korelasi negatif yang relatif lemah, yaitu −0,134, dengan p-value 0,044. Hasil tersebut signifikan secara statistik pada taraf 5% apabila asumsi pengujian terpenuhi, tetapi kekuatan hubungannya kecil. Korelasi negatif ini tidak serta-merta menunjukkan bahwa ENSO menghambat MHW, karena hubungan tersebut dapat dipengaruhi oleh faktor oseanografi regional maupun variabilitas iklim lainnya.

Dengan demikian, berdasarkan koefisien korelasi yang tersedia, monsun tampak memiliki hubungan yang lebih kuat dengan MHW dibandingkan ENSO. Akan tetapi, penetapan monsun sebagai faktor dominan secara independen memerlukan pengujian lanjutan, seperti regresi multivariat dan korelasi parsial. Perlu diperhatikan pula bahwa p-value untuk korelasi monsun tidak diberikan, sehingga signifikansi statistik kedua hubungan belum dapat dibandingkan secara lengkap.
