# **Komponen Prompt Work Specification**

**R — Role (Peran)**

Menentukan identitas atau sudut pandang AI.

**Contoh:** Dosen, peneliti, analis data, editor.

**T — Task (Tugas)**

Menjelaskan pekerjaan yang harus dilakukan AI.

**Contoh:** Menjelaskan, menganalisis, merangkum, membuat PPT.

**C — Context (Konteks)**

Memberikan informasi latar belakang agar jawaban lebih relevan.

**Contoh:** Audiens mahasiswa, topik geofisika, durasi presentasi 10 menit.

**I — Input**

Data atau informasi yang akan diproses.

**Contoh:** Artikel, dataset, gambar, kode program.

**K — Constraints (Batasan)**

Aturan yang harus dipatuhi dalam output.

**Contoh:** Maksimal 300 kata, tanpa tabel, bahasa formal.

**O — Output**

Hasil akhir yang diinginkan.

**Contoh:** Ringkasan, artikel, kode, laporan, presentasi.

**Q — Quality Criteria (Kriteria Kualitas)**

Standar untuk menilai kualitas output.

**Contoh:** Akurat, jelas, relevan, logis.

**V — Optional Verification (Verifikasi Opsional)**

Pemeriksaan tambahan untuk memastikan kualitas hasil.

**Contoh:** Fact-check, validasi data, proofreading.
